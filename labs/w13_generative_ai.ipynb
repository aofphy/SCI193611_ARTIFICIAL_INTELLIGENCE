## 1) Setup


```python
# Requires internet + GPU. Run on Google Colab with a GPU runtime.
# !pip install -q diffusers transformers accelerate
import torch
from diffusers import StableDiffusionPipeline
```

## 2) Text-to-image (Stable Diffusion)


```python
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
).to("cuda")

image = pipe("a watercolor painting of a robot studying AI textbooks").images[0]
image.save("w13_output.png")
image
```

## 3) RLHF (conceptual)


```python
# RLHF — conceptual outline (no training run here)
# 1) Supervised fine-tuning (SFT): train base LLM on demonstrations.
# 2) Reward model (RM): train a model to score outputs from human preference pairs.
# 3) Policy optimization (PPO): fine-tune the SFT model to maximize RM reward,
#    with a KL penalty against the SFT model to stay close to natural language.
#
# Libraries to explore: trl (Transformer Reinforcement Learning), trlx.
print("RLHF = SFT -> Reward Model -> PPO (KL-regularized policy optimization)")
```
